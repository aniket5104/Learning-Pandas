# Problems in Vectorized string operations in python:

```python
b= ['cat', 'dog',None, 'fish']
[i.startwith('c') for i in b]
```

1. Python cant handle Nonevalues in vectorized operations.
2. list comprehensions are relatively very slow.

---
# How pandas solve this problem.

```python
s= pd.Series(['cat', 'dog', None, 'fish'])
s.str.startswith('c')
# .str is called string Accessor
# using str we can apply any string operations on a series
```

This is Faster than list comprehensions.

## Using **`str.strip`:**
- It removes all leading and trailing spaces from the string.
- e.g.

```python
'         Aniket      '.strip()
```

```python
df['Name'].str.strip()
```

## Using **`str.split`:**

Let's create three new rows for title, first name and last name.

```python
df['last_name']= df['Name'].str.split(',').str.get(0)
df
```

```python
df[['Title', 'firstname']]= df['Name'].str.split(',').str.get(1).str.strip().str.split(' ', n=1, expand=True)
df
```

```python
df['Title'].value_counts()
```

```python
df['Title']= df['Title'].str.replace('Ms', 'Miss')
df['Title']= df['Title'].str.replace('Mlle', 'Miss')
df['Title'].value_counts()
```

## **Filtering** Using String Operations:

Find the no. of passengeres whose name start with A.

```python
df[df['firstname'].str.startswith('A')]
```

```python
df['firstname'].str.endswith('a')
df['firstname'].str.isalpha()
```

### Advance Level Filterings with **Regular Expression:**

1. **`.contains():`**

Find how many names have john as their first or last name.

```python
df[df['Name'].str.contains('john', case=False)].shape[0]
```

Find last names start and end with vowels.

```python
df[df['last_name'].str.contains('^[^aeiouAEIOU].+[^aeiouAEIOU]$')]
```
